## 1. 香港貿易特徵

- **轉口貿易為主（Re-export）：** 佔總貿易約 98%
- **服務貿易（金融、物流、專業服務）比重高：** 佔 GDP 約 90%
- **與內地貿易關係密切（CEPA）：** 內地係香港最大貿易伙伴（2025 佔總貿易約 50%）
- **全球最自由經濟體：** Heritage Foundation Index of Economic Freedom 連續多年 #1

## 2. 數字轉型對香港國際貿易的影響

### 2.1 AI + 平台經濟如何改變傳統貿易模式？

- **B2B 平台：** Alibaba 1688 + Global Sources 改變採購流程
- **B2C 平台：** Amazon + eBay 提供直接跨境零售
- **AI 客服：** 24/7 多語言支援，降低中小企進入門檻
- **智能物流：** 自動化倉儲 + 無人機配送（Amazon Prime Air）
- **預測分析：** AI 預測需求，優化庫存管理

### 2.2 Web3 / 加密資產是否構成新的「貿易項目」？

| 以太幣 (ETH) | 屬資產 | SFC 監管 |

| 區塊鏈服務 | 屬服務貿易 | 一般商業 |

### 2.3 香港在數字貿易中的優勢與挑戰

**優勢：**
- 國際金融中心 + 健全法律體系
- 對 Web3 / 加密資產友善
- 連接中西方市場
- 人才匯聚（普通法 + 中英雙語）

**挑戰：**
- 地緣政治影響（中美摩擦）
- 內地數據跨境流動限制
- 與新加坡競爭加劇
- 樓價高企影響營商成本
