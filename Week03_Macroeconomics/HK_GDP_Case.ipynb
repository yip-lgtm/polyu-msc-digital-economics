## 📈 AD 右移的主要驅動因素（2025–2026）

1. **出口強勁**：電子產品、AI 相關產品需求
2. **內需回穩**：消費券效應、樓市氣氛改善
3. **旅遊業復蘇**：內地及國際旅客回流
4. **金融服務業活躍**：跨境金融活動增加

## 🏭 SRAS 情況

- **勞工短缺 + 地緣政治風險** → SRAS 相對陡峭
- **油價上升** 對通脹有輕微向上壓力
- **香港能源依賴度低**，影響有限

## 📊 簡單 Python 練習（可選）

如果想用 Python 計算 AD-AS 數值，可以跑以下 code cell：

````python
# HK GDP 簡單計算（示意）
import matplotlib.pyplot as plt
import numpy as np

# AD-AS 模型示意
Y = np.linspace(0, 100, 100)
P_AD = 100 - Y  # AD: 向下傾斜
P_SRAS = 20 + 0.5 * Y  # SRAS: 向上傾斜（陡峭）
P_LRAS = 60  # LRAS: 垂直

plt.figure(figsize=(8, 6))
plt.plot(Y, P_AD, label='AD', color='blue', linewidth=2)
plt.plot(Y, P_SRAS, label='SRAS', color='red', linewidth=2)
plt.axvline(x=50, color='green', linestyle='--', label='LRAS')
plt.xlabel('Output (Y)')
plt.ylabel('Price Level (P)')
plt.title('Hong Kong AD-AS Model (Simplified)')
plt.legend()
plt.grid(True)
plt.show()
````
