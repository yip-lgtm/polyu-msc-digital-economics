## 1. 最新數據（2026年6月更新）

- 2025全年實質GDP增長：約 +3.2%
- 2026 Q1 實質GDP增長：+5.9%（YoY）
- 2026全年預測：2.5% – 3.5%
- 通脹（Composite CPI）：預測 2.5% – 2.6%
- 失業率：維持低位，約 3.0% – 3.3%

## 2. AD 右移的主要驅動因素（2025–2026）

香港經濟目前主要受到 Aggregate Demand (AD) 右移推動：

1. 出口強勁（尤其是電子產品、AI相關設備及半導體相關產品）
2. 內需回穩（消費券效應 + 樓市氣氛改善）
3. 旅遊業復蘇（內地及國際旅客明顯回升）
4. 金融服務業活躍（跨境金融活動增加）
5. 政府基建及開支持續支持

## 3. SRAS 情況分析

- 勞工短缺問題持續 → SRAS 相對陡峭
- 地緣政治風險及全球供應鏈不穩 → 企業成本上升壓力
- 油價上升對香港通脹影響有限（香港能源依賴度低）
- 目前未出現明顯的 成本推動型通脹（Cost-push inflation）

## 4. 你的分析（已完成版）

### Q1. 香港經濟目前處於 AD-AS 模型的哪個區域？

答案：
目前香港經濟處於 需求回升但未達過熱 的區域。

- AD 曲線正在右移（出口 + 消費 + 旅遊帶動）
- SRAS 相對陡峭（勞工短缺）
- 短期均衡點：**P* 輕微上升 + Y* 上升**，但仍未明顯超過 LRAS
- 屬於 復蘇後期 + 温和擴張 階段，尚未出現明顯的通脹缺口（Inflationary Gap）

### Q2. 2026年通脹壓力主要來自哪一方（AD 還是 AS）？

答案：
目前通脹壓力主要來自 AD 右移（需求拉動型通脹），而非 AS。

- 出口及消費需求回升是主要驅動力
- SRAS 雖然因勞工短缺而較陡，但尚未出現大規模成本上升
- 若未來油價或全球供應鏈進一步惡化，則可能轉為混合型通脹壓力

### Q3. 如果美國減息，會如何影響香港的 AD 曲線？

答案：
美國減息會令香港 AD 曲線進一步右移。

原因：
- 港元與美元掛鈎 → 香港利率會跟隨下降
- 利率下降 → 投資（I）增加 + 消費（C）增加
- 資金流入香港 → 資產價格上升 → 財富效應 → 消費進一步增加
- 整體 AD 右移，短期內對 GDP 增長有正面支持，但亦可能推高樓價及資產泡沫風險

### Q4. 對香港2026全年增長 2.5–3.5% 的預測，你是否同意？為什麼？

答案：
同意，但認為增長中後期可能略為放緩。

支持理由：
- 出口及旅遊業復蘇動能仍在
- 政府消費券及基建開支仍有支持

風險因素：
- 全球經濟放緩（尤其是歐美需求減弱）
- 地緣政治風險升溫
- 樓市復蘇力度若低於预期，內需會被拖累

結論：2.8% – 3.3% 為較合理預測區間，屬温和復蘇而非強勁反彈。

## 5. 總結

2025–2026年香港經濟處於 AD 驅動的復蘇階段，短期內 GDP 增長及通脹均會温和上升。
政府應密切監察資產價格及勞工市場，避免需求過熱轉化為資產泡沫或成本推動型通脹。
