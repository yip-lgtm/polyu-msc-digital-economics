## 1. 香港聯繫匯率制度簡介

- 港元與美元掛鈎（**1 USD = 7.75–7.85 HKD**）
- **Currency Board** 制度：每發行 7.80 港元，必須有 1 美元儲備支持
- HKMA 主要透過 **利率調整** 和 **流動性管理** 維持匯率穩定
- 香港幾乎沒有獨立的貨幣政策（利率跟隨美國）

### 聯繫匯率制度特點

| 特點 | 描述 |
|------|------|
| **匯率穩定** | 港元/美元匯率波動極小 |
| **無獨立貨幣政策** | 香港利率 = 美國利率 ± 息差 |
| **自動調節** | 資金流入/流出自動調整利率 |
| **外匯基金** | 充足美元儲備支持 |
| **歷史** | 1983 年 10 月設立至今 40+ 年 |

## 4. Crowding-out Effect 詳解

### 4.1 定義
政府增加開支（G↑）→ 需要融資（發債 / 加稅）→ 利率上升 → 私人投資（I↓）

### 4.2 香港情況
- 香港政府有龐大財政儲備 → 發債規模有限
- 香港私人借貸市場活躍 → 影響相對較小
- **結論：香港的 Crowding-out Effect 不太明顯**

### 4.3 嚴重情況
- 政府赤字嚴重（> GDP 5%）
- 經濟接近/已達充分就業
- 利率已經很高

## 6. 總結

### 6.1 香港獨特之處

| 範疇 | 工具 | 限制 |
|------|------|------|
| 貨幣政策 | 跟隨美國利率 | 無獨立空間 |
| 財政政策 | 預算案（消費券、基建、稅務寬減）| 視乎財政狀況 |
| 匯率政策 | 7.75-7.85 港元掛鈎 | 不可調整 |
| 監管政策 | 樓市辣招、宏觀審慎 | 行政主導 |

### 6.2 政策組合建議

- **逆周期財政政策：** 經濟低迷時派消費券、增基建；經濟過熱時減少開支
- **結構性改革：** 增加房屋供應、培訓人才、推動創科
- **聯繫匯率維持：** 保持金融穩定（雖然犧牲貨幣政策獨立性）

### 6.3 未來觀察重點

- 美國息口走向（影響香港 HIBOR）
- 中美關係及地緣政治
- 香港樓市復蘇力度
- 政府財政赤字持續性
