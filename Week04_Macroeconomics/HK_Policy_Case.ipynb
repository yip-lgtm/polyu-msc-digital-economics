## 1. 香港聯繫匯率制度（Linked Exchange Rate System, LERS）

**基本資料：**
- 港元與美元掛鈎範圍：**7.75 – 7.85 HKD/USD**
- 1983年10月17日設立
- 負責機構：**香港金融管理局（HKMA）**
- 政策目標：維持港元匯率穩定（金融穩定）

**強方兌換保證：** 7.75 HKD/USD
**弱方兌換保證：** 7.85 HKD/USD

**運作機制：**
- 港元偏強（接近 7.75）：HKMA 買入美元、賣出港元 → 銀行體系結餘增加 → 利率下降
- 港元偏弱（接近 7.85）：HKMA 賣出美元、買入港元 → 銀行體系結餘減少 → 利率上升

## 2. HKMA 政策工具

| 工具 | 描述 | 效果 |
|------|------|------|
| **貼現窗** | 銀行向 HKMA 借隔夜錢 | 利率下限 |
| **外匯基金票據（EF Bills）** | HKMA 發行短債 | 調節銀行流動性 |
| **總結餘** | 銀行體系結餘總額 | 影響 HIBOR |
| **LERS 自動調節** | 強弱方保證觸發 | 自動利率調整 |

**重點：香港沒有獨立貨幣政策**，利率主要跟隨美國聯邦基金利率。

## 4. 美國息口對香港的傳導機制

### 4.1 利率傳導路徑

```
美聯儲減息
    ↓
HIBOR 下降（聯繫匯率）
    ↓
1. 借貸成本下降 → I↑（投資增加）
    ↓
2. 樓按利率下降 → 樓市需求↑ → 財富效應 → C↑
    ↓
3. 港元偏弱（資金流出）→ 出入口改善
    ↓
AD 右移 + AS 改善 → GDP↑
```

### 4.2 量化分析

- 美國減息 0.25% → 香港 HIBOR 下降約 0.20–0.25%
- 樓按利率每降 0.25% → 樓市成交量上升約 5–8%
- AD 短期增加約 0.3–0.5%
