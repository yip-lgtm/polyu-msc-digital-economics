## 3. 數據價值化案例

### 3.1 香港政府「智方便」

- 2022 年推出，註冊用戶 400 萬+

## 4. 平台壟斷與反壟斷監管

### 4.1 香港監管框架

- **《競爭條例》** （2015）
- **競爭事務委員會（Competition Commission）**
- 兩大禁止行為：
  - **第一行為守則：** 反競爭協議
  - **第二行為守則：** 濫用市場權勢

### 4.2 內地 vs 香港監管比較

| 範疇 | 內地 | 香港 |
|------|------|------|
| 監管機構 | 國家市場監督管理總局 | 競爭事務委員會 |
| 執法力度 | 高（2020-2021 風暴）| 中（個別案例）|
| 罰款 | 可達年營業額 4-10% | 最高 HK$____ |
| 跨境協調 | 中央集權 | 普通法 + 國際協調 |
| Web3 | 嚴格 | 友善 |
