## 4. 數據價值化現況

### 4.1 香港在數據開放（Open Data）方面的進度

- **data.gov.hk：** 超過 4,000 個數據集
- **API 經濟：** 政府部門開放 API
- **跨部門數據共享：** Common Data Framework
- **國際比較：** 香港 Open Data 排名約 30+ 名（落後於新加坡、英國）

### 4.2 PDPO（個人資料私隱條例）對數據 monetization 的限制

**6 大原則：**
1. 收集目的明確
2. 準確性
3. 使用限制
4. 保安
5. 透明度
6. 資料當事人參與

**影響：**
- 跨境數據流動受限
- 用戶同意要求嚴格
- 罰款最高 HK$100 萬 + 民事索償
- 2021 年修訂加強執法

### 4.3 Web3 項目如何合法進行數據/資產 tokenization

- **SFC VASP 牌照**（2023 生效）
- **穩定幣條例**（2025 生效）
- **現貨 ETF**（2024 批准）
- **沙盒機制**（Fintech 試點）
- **代幣化資產**（綠債、基金、房地產）
